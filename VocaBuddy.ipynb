```python
# VocaBuddy 單字庫初始化與擴充
vocabulary = {
    "apple": "蘋果",
    "orange": "橘子",
    "banana": "香蕉"
}

print(f"目前單字庫共有 {len(vocabulary)} 個單字。")
for word, meaning in vocabulary.items():
    print(f"- {word}: {meaning}")
```

```python
import random

def quiz_vocabulary(vocab_dict):
    """
    隨機從單字庫中抽取單字並詢問使用者中文意思
    """
    if not vocab_dict:
        print("單字庫是空的！")
        return

    # 隨機挑選一個單字
    word = random.choice(list(vocab_dict.keys()))
    correct_meaning = vocab_dict[word]

    # 詢問使用者
    user_input = input(f"請問 {word} 的中文意思是什麼？ ").strip()

    # 判斷對錯
    if user_input == correct_meaning:
        print("great")
    else:
        print(f"wrong (正確答案是: {correct_meaning})")

# 呼叫函式進行測驗
quiz_vocabulary(vocabulary)
```
